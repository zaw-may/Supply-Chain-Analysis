```python
import pandas as pd

file_url = "..\Sample Datasets\DataCoSupplyChainDataset1.csv"
chunk_size = 5000
output_file_url = "..\Sample Datasets\SupplyChainDataset_"

for i, chunk in enumerate(pd.read_csv(file_url, chunksize = chunk_size)):
    output_file = f"{output_file_url}{i+1}.csv"
    chunk.to_csv(output_file, index = False)
    print(f"Saved {output_file}")
```
